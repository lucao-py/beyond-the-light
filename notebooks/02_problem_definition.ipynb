```python
import pandas as pd
```

```python
# carregando df
df = pd.read_csv("../data/raw/kepler_cumulative.csv")
```

```python
df = df.copy()
```

```python
# separando os exoplanetas candidatos
df_candidates = df[df['koi_disposition'] == 'CANDIDATE']

# removendo candidates do meu df
df = df[df['koi_disposition'] != 'CANDIDATE']
```

```python
df["target"] = df["koi_disposition"].map({
    "CONFIRMED": 1,
    "FALSE POSITIVE": 0
})

df["target"].value_counts()
```

```python
# colunas descartadas
leakage_cols = [
    "kepler_name",
    "koi_disposition",
    "koi_pdisposition",
    "koi_score",
    "koi_comment",
    "koi_disp_prov",
    "koi_fpflag_nt",
    "koi_fpflag_ss",
    "koi_fpflag_co",
    "koi_fpflag_ec"
]

identifier_cols = [
    "kepid",
    "kepoi_name"
]

metadata_cols = [
    "koi_delivname",
    "koi_vet_stat",
    "koi_vet_date",
    "koi_tce_plnt_num",
    "koi_tce_delivname",
    "koi_datalink_dvs",
    "koi_datalink_dvr",
    "koi_parm_prov",
    "koi_sparprov",
    "koi_fittype",
    "koi_limbdark_mod",
    "koi_trans_mod"
]

review_cols = [
    "ra",
    "ra_err",
    "ra_str",
    "dec",
    "dec_err",
    "dec_str",
    "koi_count",
    "koi_quarters"
]

review_cols += [
    col for col in df.columns
    if col.startswith(("koi_fwm_", "koi_dicco_", "koi_dikco_"))
]
```

```python
excluded_cols = (
    leakage_cols
    + identifier_cols
    + metadata_cols
    + review_cols
    + ["target"]
)

feature_cols = [
    col for col in df.columns
    if col not in excluded_cols
]

X = df[feature_cols].copy()
y = df["target"].copy()
groups = df["kepid"].copy()

print("Total de features elegíveis:", len(feature_cols))
```
